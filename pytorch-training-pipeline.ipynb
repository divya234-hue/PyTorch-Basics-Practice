```python
class MySimpleNN():
    def __init__(self, x):
        self.weight = torch.rand(
            x.shape[1], 1,
            dtype=torch.float64,
            requires_grad=True
        )
        self.bias = torch.zeros(
            1,
            dtype=torch.float64,
            requires_grad=True
        )

    def forward(self, x):
        z = torch.matmul(x, self.weight) + self.bias
        y_pred = torch.sigmoid(z)
        return y_pred

    def loss_function(self, y_pred, y):
        epsilon = 1e-7
        y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

        loss = -(y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred))
        return loss.mean()
```

```python
learning_rate = 0.1
epochs = 15
```

learning rate = modelko btata training time parameters/weight ko kitna change krna hai 
epochs = Model poore dataset ko kitni baar dekhega


```python
model = MySimpleNN(x_train_tensor)

for epoch in range(epochs):

    # forward pass
    y_pred = model.forward(x_train_tensor)

    # loss calculate
    loss = model.loss_function(y_pred, y_train_tensor)
    print(f'Epoch: {epoch}, Loss: {loss.item()}')

    # backward pass
    loss.backward()

    # updating parameters
    with torch.no_grad():
        model.weight -= learning_rate * model.weight.grad
        model.bias -= learning_rate * model.bias.grad

    # zero gradients
    model.weight.grad.zero_()
    model.bias.grad.zero_()
    #prnt loss in each epochs
    print(f'Epochs:{epochs+1},loss:{loss.item()}')
```
