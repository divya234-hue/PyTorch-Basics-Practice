```python
import torch
import torch.nn as nn
model = nn.Linear(1,1)
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(),lr=0.01)
x = torch.tensor([[1.0],[2.0],[3.0]])
y = torch.tensor([[2.0],[4.0],[6.0]])
for epoch in range(100):
    prediction = model(x)
    loss = loss_fn(prediction,y)
    loss.backward()
    optimizer.step()
    optimizer.zero_grad()

print(model(torch.tensor([[4.0]])))
```
